## Introduction to astropy units

First, load the relevant python packages.

```python
import numpy as np # Array handling
from matplotlib import pyplot as plt # Plotting
from astropy import units # Units handling
from ipywidgets import interact, FloatSlider
```

In order to avoid confusion with units, we strongly recommend to use the astropy.units package which "attaches" units to your floats/arrays. As an example, one can define the gravitational constant as follows:

```python
G = 6.6743e-11 * units.m**3 / units.kg / units.s**2 # Gravitational constant in the proper units
```

These units are properly converted when performing mathematical operations, and also shown when printed:

```python
print(G)
print(G**2)
```

You can also convert to other units:

```python
print(G.to('pc3/(Msun*Gyr2)'))
```

Note: astropy also has a constants package which contains many predefined constants - you can also use those instead of defining them yourself.

```python
from astropy import constants
print(constants.G)
```

Another example involving numpy arrays:

```python
t = np.linspace(0., 100., 11) * units.s # Create a numpy array from 0 to 100 seconds, in 10 equal-sized bins

def x(v, t): # Define a function for the position as a function of velocity
    x = v * t 
    return x

v = 5. * units.m / units.s # Define the velocity

print(x(v, t)) # Outputs the position as a function of time, with the correct unit
```

## Task 1

To get started, consider the spherically symmetric Hernquist potential which has the following density profile: $\Large \rho(r)=\frac{2M}{4\pi b^3}\frac{b}{r}(1+\frac{r}{b})^{-3}$, where $M$ is the total mass in the system and $b$ a scale parameter. From the density profile, derive the following quantities:
- Mass enclosed within radius $M(r)$
- Potential $\Phi(r)$
- Circular velocity $v_\mathrm{circ}(r)$
- Escape velocity $v_\mathrm{esc}(r)$
Note: The equation to derive the escape velocity is not in the syllabus, but you can derive it using energy conservation.

To perform the calculations, feel free to use any resource (e.g. WolframAlpha). Once you have computed these quantities, plot them together with the density as a function of radius. What happens when $b\rightarrow0$?

The rotation curve has a global maximum. Derive the radius (as a function of the parameters in the potential $M$ and $b$) where the rotation curve has its maximum, and check whether the plot of your rotation curve is consistent with this.

\begin{equation}
\rho(r)=\frac{2M}{4\pi b^3}\frac{b}{r}(1+\frac{r}{b})^{-3}
\end{equation}

\begin{equation}
M(r) = \frac{Mr^2}{(r+b)^2}
\end{equation} 

\begin{equation}
\Phi(r) = \frac{-GMr}{(r+b)^2} - \frac{GMb}{(r+b)^2}
\end{equation} 

\begin{equation}
v_{circ} = \sqrt{\frac{GMr}{(r+b)^2}}
\end{equation}

```python
def density(r,M,b):
    r = r.to(units.kpc)          # ensure r is in kpc
    b = b.to(units.kpc)# ensure b is in kpc
    M = M.to(units.Msun)         # ensure M is in Msun
    return (2*M/(4*np.pi*b**3))*(b/r)*(1+r/b)**3
    
def Mass(r,b):
    r = r.to(units.kpc)          # ensure r is in kpc
    b = b.to(units.kpc)          # ensure b is in kpc
    return M*r**2/(r+b)**2

def Phi(r,b,M):
    r = r.to(units.kpc)          # ensure r is in kpc
    b = b.to(units.kpc)          # ensure b is in kpc
    M = M.to(units.Msun)         # ensure M is in Msun
    return -G*M/(r+b)**2 * (b+r)

def v_circ(r,b,M):
    r = r.to(units.kpc)          # ensure r is in kpc
    b = b.to(units.kpc)          # ensure b is in kpc
    M = M.to(units.Msun)         # ensure M is in Msun
    return np.sqrt(G*M*r/(r+b)**2)

def v_esc(r,b,M):
    return np.sqrt(-2*Phi(r,b,M)/M)
```

```python
r = np.linspace(0,50)*units.kpc
b = 10*units.kpc
M = 10**10*units.Msun
colors = ['red', 'blue', 'green', 'purple', 'orange']

fig, axs = plt.subplots(3,2, figsize = (7,7))

# Plot each function in a separate subplot
axs[0, 0].plot(r, density(r, M, b), color = colors[0])
axs[0, 0].set_title(r'$\rho(r)$')
axs[0, 0].set_xlabel('r')
axs[0, 0].set_ylabel(r'$\rho$')

axs[0, 1].plot(r, Mass(r, b), color = colors[1])
axs[0, 1].set_title(r'$M(r)$')
axs[0, 1].set_xlabel('r')
axs[0, 1].set_ylabel(r'$M$')

axs[1, 0].plot(r, Phi(r,b,M), color = colors[2])
axs[1, 0].set_title(r'$\Phi(r)$')
axs[1, 0].set_xlabel('r')
axs[1, 0].set_ylabel(r'$\Phi$')

axs[1, 1].plot(r, v_circ(r,b,M), color = colors[3])
axs[1, 1].set_title(r'$v_{circ}$')
axs[1, 1].set_xlabel('r')
axs[1, 1].set_ylabel(r'$v_{circ}$')

axs[2, 0].plot(r, v_esc(r,b,M), color = colors[4])
axs[2, 0].set_title(r'$v_{esc}$')
axs[2, 0].set_xlabel('r')
axs[2, 0].set_ylabel(r'$v_{esc}$')

# Add layout adjustments
for ax in axs.flat:
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Create interactive function
def update_plots(b):
    fig, axs = plt.subplots(3, 2, figsize=(7, 7))

    axs[0, 0].plot(r, density(r, M, b), color=colors[0])
    axs[0, 0].set_title(r'$\\rho(r)$')
    axs[0, 0].set_xlabel('r')
    axs[0, 0].set_ylabel(r'$\\rho$')

    axs[0, 1].plot(r, Mass(r, b), color=colors[1])
    axs[0, 1].set_title(r'$M(r)$')
    axs[0, 1].set_xlabel('r')
    axs[0, 1].set_ylabel(r'$M$')

    axs[1, 0].plot(r, Phi(r, b, M), color=colors[2])
    axs[1, 0].set_title(r'$\\Phi(r)$')
    axs[1, 0].set_xlabel('r')
    axs[1, 0].set_ylabel(r'$\\Phi$')

    axs[1, 1].plot(r, v_circ(r, b, M), color=colors[3])
    axs[1, 1].set_title(r'$v_{circ}$')
    axs[1, 1].set_xlabel('r')
    axs[1, 1].set_ylabel(r'$v_{circ}$')

    axs[2, 0].plot(r, v_esc(r, b, M), color=colors[4])
    axs[2, 0].set_title(r'$v_{esc}$')
    axs[2, 0].set_xlabel('r')
    axs[2, 0].set_ylabel(r'$v_{esc}$')

    for ax in axs.flat:
        ax.grid(True)

    plt.tight_layout()
    plt.show()

# Add slider
interact(update_plots, b=FloatSlider(value=10, min=1, max=50, step=1, description='b (kpc)'))
```

## The galpy package

For more complex potentials, especially ones without spherical symmetry, these calculations become a lot more cumbersome - often even without analytical expressions. Fortunately, you don't have to numerically solve integrals and ODE's for the rest of this course: The galpy package has you covered! This is a python package which does all the hard work of calculating potentials, rotatition curves, orbits and much more for you. For the rest of this exercise we'll use galpy.

An important note about units in galpy: galpy has an internal unit system which is not very intuitive to use. Hence, we strongly suggest to consistently work with astropy units. By default, galpy works with those units, but does not output quantities including units. To change this, go to the galpy settings file and change "astropy-units" to True. See the "Configuration file" section at the end of the following website on how to find this galpy settings file: https://docs.galpy.org/en/v1.11.0/installation.html#tl-dr

First, verify whether galpy works and outputs units properly:

```python
from galpy.potential import PlummerPotential

print(PlummerPotential(amp = 1e10 * units.Msun, b = 10. * units.kpc)(0. * units.kpc, 0. * units.kpc))
```

If everything is working properly, this should output the central value of the spherically symmetric Plummer potential with $M=10^{10}\,\mathrm{M}_\odot$ and $b=10\,\mathrm{kpc}$ with the proper unit, which is $\Phi(0)=-4300.917\,\mathrm{km^2}\mathrm{s}^{-2}$. Note: If you call the exact same function but don't add units to the arguments, you'll get a completely different result (without units) because of galpy's internal unit system:

```python
print(PlummerPotential(1e10, 10.)(0. * units.kpc, 0. * units.kpc))
```

Hence, when calling galpy functions, always use the appropriate units!

With galpy, we can also load some more complicated potentials such as the axisymmetric Miyamoto-Nagai-potential:

```python
from galpy.potential import MiyamotoNagaiPotential
```

This potential has three parameters, the total mass $M$ and two length scales $a$ and $b$. You can set up a Miyamoto-Nagai-potential specifying these parameters as follows:

```python
potential = MiyamotoNagaiPotential(amp = 1e10 * units.Msun, a = 10. * units.kpc, b = 2 * units.kpc) # The amp parameter corresponds to the total mass
```

Now, you can access various quantities such as the potential or the density as a function of the cylindrical coordinates $R$ and $z$:

```python
print('Potential:', potential(10. * units.kpc, 0. * units.kpc))
print('Density:', potential.dens(2. * units.kpc, -3 * units.kpc))
```

Additionally, you can directly visualize the density distribution corresponding to the Miyamoto-Nagai-potential:

```python
potential.plotDensity(cmap = 'viridis')
plt.show()
```

This functionality is convenient for quick visualization, but doesn't offer that much flexibility to modify the plot. For instance, in the above plot it would be nice to have a colorbar on the side which shows what quantity and what values are plotted. To overcome this, we can create the plot ourselves as follows:

```python
R = np.linspace(0., 40., 100) * units.kpc # Grid in R-coordinate
z = np.linspace(-20., 20., 100) * units.kpc # Grid in z-coordinate
RR, zz = np.meshgrid(R, z) # Creates a 2D-grid

density = potential.dens(RR, zz) # 2D array with the mass density as a function of R and z

fig = plt.figure(figsize = (6., 5.))
ax = fig.add_subplot(111)

im = ax.pcolormesh(RR.value, zz.value, np.log10(density.to('Msun/pc3').value)) # Plot the logarithm of the density, in units of solar mass/pc^3

cbar = plt.colorbar(im, ax = ax)
cbar.minorticks_on()
cbar.set_label(r'Density $\log_{10}(\rho/(\mathrm{M}_\odot\mathrm{pc}^{-3}))$')


ax.set_xlabel(r'$R\,[\mathrm{kpc}]$')
ax.set_ylabel(r'$z\,[\mathrm{kpc}]$')

ax.minorticks_on()
ax.tick_params(which = 'both', direction = 'in', top = True, right = True)

plt.tight_layout()
plt.show()
```

This is the same plot of density, but now with the appropriate colorbar which indicates what we're plotting.

**Most importantly: When working with galpy and questions arise, check the documentation at the galpy website (https://docs.galpy.org/en/v1.11.0/index.html)** For instance, the Miyamoto-Nagai-potential is documented here: https://docs.galpy.org/en/v1.11.0/reference/potentialmiyamoto.html, and you find all available properties and functions for galpy potentials here: https://docs.galpy.org/en/v1.11.0/reference/potential.html.
