```python
import galpy
import numpy as np
import h5py
import os

#10^5 solar mass per "dark matter particle"
#Careful with units!!!
#c in beginning of units means 'comoving'
```

```python
#Tests with h5py (Header)

data  = h5py.File("cutout_17.hdf5", 'r')

print(list(data['Header'].attrs)) #No raw data in Header, but has all attributes about simulations
print(data['Header'].attrs['CutoutID']) #Should give 17

print(data['Header'].attrs['BoxSize']) #Has no units! But is a length
print(data['Header'].attrs['UnitLength_in_cm'], "\n") #The unit of length. So the boxsize is 35000 * 3.08...e+21cm

for el in list(data['Header'].attrs):
    print(f"The value of {el} is: {data['Header'].attrs[el]}")

#set hubble parameter
h = data['Header'].attrs['HubbleParam'] #useful later to get the right units
```

```python
#Tests with h5py (PartType1)

data['PartType1'].keys()

for el in list(data['PartType1'].keys()):
    print(f"The value of {el} is: {data['PartType1'][el]}")
print("\n")

print(data['PartType1']['ParticleIDs']) # shape (, ) is 1D
print(data['PartType1']['Coordinates']) #shape (, 3) is 3D (coordinates xyz)

# Don't use data['PartType1']['Potential'] since it also includes other halos.
# Derive it from other raw data

print(list(data['PartType1']['Coordinates'].attrs)) #attributes of Coordinates (also for other keys obviously)
data['PartType1']['Coordinates'].attrs['a_scaling'] #see the attribute a_scaling for key "Coordinates"
```
