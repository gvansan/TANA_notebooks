## Test the galpy installation

```galpy``` is a python package which performs numerical calculations of potentials and orbits. Many of these calculations use ```C``` code under the hood of ```python```. If ```galpy``` is not installed correctly, these calculations can fail. Here we test a few ```galpy``` calculations to see whether everything works properly. No worries, you don't need to understand yet how these calculations work - you'll learn that during the coming weeks in the lectures and exercise sessions.

Important note: ```galpy``` uses an internal unit system which is not very intuitive to use. Hence, after installing ```galpy```, change the unit system to physical units by editing the .galpyrc file in your home directory.

````python
from galpy.potential import MiyamotoNagaiPotential, AnySphericalPotential
from galpy.orbit import Orbit
from astropy import units
import numpy as np
````

````python
orbit = Orbit([5. * units.kpc, 0. * units.km / units.s, 10. * units.km / units.s, 1. * units.kpc, 0 * units.km / units.s, 0. * units.rad])
potential = MiyamotoNagaiPotential(amp = 1e10 * units.Msun, a = 10. * units.kpc, b = 1. * units.kpc)

print('Energy of the orbit:', orbit.E(pot = potential))
print('Radial period of the orbit:', orbit.Tr(analytic = True, pot = potential))
````

````python
orbit = Orbit([5. * units.kpc, 0. * units.km / units.s, 10. * units.km / units.s, 0.1 * units.kpc, 0 * units.km / units.s, 0. * units.rad])
ts = np.linspace(0. * units.Gyr, 1. * units.Gyr, 10000)

orbit.integrate(ts, potential)
````

````python
def densityFunction(r):

    return 1e10 * units.Msun / units.kpc**3 / (1. + r / units.kpc)**3

orbit = Orbit([5. * units.kpc, 0. * units.km / units.s, 0.01 * units.km / units.s, 1. * units.kpc, 0 * units.km / units.s, 0. * units.rad])
potential = AnySphericalPotential(dens = densityFunction)

print('Energy of the orbit:', orbit.E(pot = potential))
````

This specific potential does not support `C` integration, which is why you should get a galpyWarning here. But the code should still finish.

````python
orbit = Orbit([5. * units.kpc, 0. * units.km / units.s, 10. * units.km / units.s, 0.1 * units.kpc, 0 * units.km / units.s, 0. * units.rad])
ts = np.linspace(0. * units.Gyr, 0.01 * units.Gyr, 100)

orbit.integrate(ts, potential)
````
