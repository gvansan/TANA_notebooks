## Introduction to galpy distribution functions

The calculations you performed in task 1 become a lot more cumbersome (often even impossible to do analytically) for more realistic models. Once again, galpy can help you out by providing a lot of functionality related to DFs for many different potentials. To show an example of how to work with galpy DFs, we'll consider the isotropic Plummer DF. Import the needed packages:

```python
from galpy.df import isotropicPlummerdf
from galpy.potential import PlummerPotential
from astropy import units
```

First, set up a galpy Plummer potential object:

```python
M_galpyExample = 1e10 * units.Msun
b_galpyExample = 1. * units.kpc
potential_galpyExample = PlummerPotential(amp = M_galpyExample, b = b_galpyExample)
```

The corresponding (isotropic) DF can now easily be defined as follows:

```python
df_galpyExample = isotropicPlummerdf(pot = potential_galpyExample)
```

Now, you can access various properties of the distribution function (velocity dispersions, anisotropies, moments of the DF) as follows:

```python
r_galpyExample = 1. * units.kpc # Example radius

print('Radial velocity dispersion:', df_galpyExample.sigmar(r_galpyExample))
print('Tangential velocity dispersion:', df_galpyExample.sigmat(r_galpyExample)) # Is higher than the radial velocity dispersion by a factor of sqrt(2)
print('Anisotropy:', df_galpyExample.beta(r_galpyExample)) # Zero because this is an isotropic DF
```
